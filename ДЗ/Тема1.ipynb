### Пример 1.8
> В конкурсе по 5 номинациям участвуют 10 кинофильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены: а) различные призы; б) одинаковые призы?

* Рассмотрим событие А - различные призы

$P(A) = n^m = {10^5} = 100000$

```python
pow(10, 5)
```

* Рассмотрим событие B - одинаковые призы

$P(B) = C_{n+m-1}^m = C_{10+5-1}^5 = C_{14}^5 = \frac{14 \cdot 13 \cdot 12 \cdot 11 \cdot 10}{5!} = 2002$

```python
round(sc.comb(14, 5), 4)
```

### Пример 1.9
> Сколько существует семизначных чисел, состоящих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 - по 2 раза?

* Рассмотрим событие А - каждое семизначное число является перестановкой с повторениями из 7 элементов.

$P(A) = {P_{7} (3,2,2)} = \frac{7!}{3! \cdot 2! \cdot 2!} = 210$

```python
round(sc.factorial(7) / (sc.factorial(3) * sc.factorial(2) * sc.factorial(2)))
```

### Пример 1.10
> Буквы Т, Е, И, Я, Р, О написаны на отдельных
карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к друтой: а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

* Рассмотрим событие А - получение слова «ТОР».

$P(A) = \frac{m}{n} = \frac{1}{6 \cdot 5 \cdot 4} = \frac{1}{120}$

```python
round(1 / (6*5*4), 4)
```

* Рассмотрим событие B - получение слова «ТЕОРИЯ».

$P(B) = \frac{m}{n} = \frac{1}{6!} = \frac{1}{720}$

```python
round(1 / sc.factorial(6), 4)
```

### Пример 1.11
> Используя условие примера 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных карточках написаны три буквы А, две буквы Н и одна буква С.

* Рассмотрим событие А - получение слова «АНАНАС».

$P(A) = \frac{m}{n} = \frac{P_3 \cdot P_2}{P_6} = \frac{3! \cdot 2!}{6!} = \frac{1}{60}$

```python
round(sc.factorial(3) * sc.factorial(2) / sc.factorial(6), 4)
```

### Пример 1.16
> За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что: а) никакие два лица одного пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и
женщины образуют 5 супружеских пар.

* Рассмотрим событие А - никакие два лица одного
пола не сядут рядом.

$P(A) = \frac{m_1}{n} = \frac{2(5!)^2}{10!} = \frac{2 \cdot 120^2}{3 628 800} = 0,00794$

```python
round(2 * pow(sc.factorial(5), 2) / sc.factorial(10), 4)
```

* Рассмотрим событие B - супруги, образующие пять супружеских
пар, сядут рядом.

$P(B) = \frac{m_2}{n} = \frac{5!2^5}{10!} = \frac{120 \cdot 32}{3 628 800} = 0,00106$

```python
round(sc.factorial(5) * pow(2, 5) / sc.factorial(10), 4)
```

### Пример 1.17
> В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе.

* Рассмотрим событие А - пассажиры попали в два купе.

$P(A) = \frac{m_1}{n} = \frac{C_9^2 \cdot C_8^7}{C_{36}^7} = \frac{36 \cdot 8}{8347680} = 0,0000345$

```python
print(f"{(sc.comb(9, 7) * sc.comb(8, 7)) / sc.comb(36, 7):.5f}")
```

* Рассмотрим событие B - пассажиры попали в семь купе.

$P(B) = \frac{m_2}{n} = \frac{C_9^7 \cdot 4^7}{C_{36}^7} = \frac{36 \cdot 16384}{3 347 680} = 0,07066$

```python
round(sc.comb(9, 7) * pow(4, 7) / sc.comb(36, 7), 4)
```
