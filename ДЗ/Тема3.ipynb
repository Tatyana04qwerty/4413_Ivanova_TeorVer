### Задача 1.55
> Прибор, работающий в течение времени t, состоит из трех узлов, каждый из которых независимо от других может за это время выйти из строя. Неисправность хотя бы одного узла выводит прибор из строя целиком. Вероятность безотказной работы в течение времени t первого узла равна 0,9, второго - 0,95, третьего - 0,8. Найти вероятность того, что в течение времени t прибор выйдет из строя.

* $A_i$ - безотказная работа i-го узла
* $\bar{A}$ - прибор выйдет из строя
* $P(\bar{A}) = 1 - P(A) = 1 - P(A_1 * A_2 * A_3) = 1 - 0.9 * 0.95 * 0.8$

```python
round(1 - 0.9 * 0.95 * 0.8, 4)
```

### Задача 1.58
> Вероятность своевременного выполнения студентом контрольной работы по каждой из трех дисциплин равна соответственно 0,6, 0,5 и 0,8. Найти вероятность своевременного выполнения контрольной работы студентом: а) по двум дисциплинам; б) хотя бы по двум дисциплинам.

* $A_i$ - своевременное выполнение i-й дисциплины
* $A$ - своевременное выполнение по двум дисциплинам
* $B$ - своевременное выполнение хотя бы по двум дисциплинам

* a) $P(A) = P(A_1*A_2*\bar{A_3}) + P(A_1*\bar{A_2}*A_3) + P(\bar{A_1}*A_2*A_3)$

```python
round(0.6 * 0.5 * 0.2 + 0.6 * 0.5 * 0.8 + 0.4 * 0.5 * 0.8, 4)
```

* б) $P(B) = P(A_1*A_2*\bar{A_3}) + P(A_1*\bar{A_2}*A_3) + P(\bar{A_1}*A_2*A_3) +  P(A_1*A_2*A_3)$

```python
round(0.6 * 0.5 * 0.2 + 0.6 * 0.5 * 0.8 + 0.4 * 0.5 * 0.8 + 0.6 * 0.5 * 0.8, 4)
```

### Задача 1.61
> Среди 20 поступающих в ремонт часов 8 нуждаются в общей чистке механизма. Какова вероятность того, что среди взятых одновременно наудачу 3 часов по крайней мере двое нуждаются в общей чистке механизма?

* $P(A) = \frac{m}{n} = \frac{C_8^2 \cdot C_{12}^1 + C_8^3}{C_{20}^3}$

```python
round((sc.comb(8, 2) * sc.comb(12, 1) + sc.comb(8, 3)) / sc.comb(20, 3), 4)
```

### Задача 1.64
> В коробке 10 красных, 3 синих и 7 желтых карандашей. Наудачу вынимают 3 карандаша. Какова вероятность того, что они все: а) разных цветов; б) одного цвета?

* а) $P(A) = \frac{m}{n} = \frac{C_{10}^1 \cdot C_3^1 \cdot C_7^1}{C_{20}^3}$

```python
round((sc.comb(10, 1) * sc.comb(3, 1) * sc.comb(7, 1)) / sc.comb(20, 3), 4)
```

* б) $P(B) = \frac{m}{n} = \frac{C_{10}^3 + C_3^3 + C_7^3}{C_{20}^3}$

```python
round((sc.comb(10, 3) + sc.comb(3, 3) + sc.comb(7, 3)) / sc.comb(20, 3), 4)
```

### Задача 1.67
> Сколько раз нужно провести испытание, чтобы с вероятностью, не меньшей Р, можно было утверждать, что по крайней мере один раз произойдет событие, вероятность которого в каждом испытании равна р? Дать ответ при р = 0,4 и Р = 0,8704.

* $P(A) = 1 - (1 - p)^n$
* $0.8704 ≤ 1 - ( 1 - 0,4)^n$
* $0.6^n ≤ 0.1296$

```python
math.ceil(math.log(1 - 0.8704) / math.log(0.4))
```

### Задача 1.68
> На полке стоят 10 книг, среди которых 3 книги по теории вероятностей. Наудачу берутся три книги. Какова вероятность, что среди отобранных хотя бы одна книга по теории вероятностей?

* $P(A) = \frac{m}{n} = \frac{C_3^1 \cdot C_7^2 + C_3^2 \cdot C_7^1 + C_3^3}{C_{10}^3}$

```python
round((sc.comb(3, 1) * sc.comb(7, 2) + sc.comb(3, 2) * sc.comb(7, 1) + sc.comb(3, 3)) / sc.comb(10, 3), 4)
```

### Задача 1.71
> Радист трижды вызывает корреспондента. Вероятность того, что будет принят первый вызов, равна 0,2, второй - 0,3, третий - 0,4. События, состоящие в том, что данный вызов будет услышан, независимы. Найти вероятность того, что корреспондент услышит вызов радиста.

* $A_i$ - i-й звонок будет принят
* $P(A) = 1 - P(\bar A) = 1 - P(\bar A_1 * \bar A_2 * \bar A_3)$

```python
round(1 - 0.8 * 0.7 * 0.6, 4)
```
