### Смоделируйте вычисление геометрической вероятности (на примере из лекции про метание дротиков в игровой круг, площадь которого в два раза меньше площади стены). Показать теоретический расчет и графическую иллюстрацию для рассматриваемого примера.

Человек сидит под яблоней. Крона дерева — круг радиуса $R = 3$ м. Человек находится в точке, отстоящей от центра кроны на расстоянии $d = 1$ м. Считаем, что человек занимает площадку радиусом $r=0.5$ м. В центре кроны растёт ствол — круг радиуса $r_0 = 0.3$ м. Яблоко падает вертикально вниз из случайной точки кроны, равномерно распределённой по площади. Яблоко не может упасть на ствол.
Найти вероятность того, что яблоко упадёт на человека.

$S_{кроны} = πR^2 = 9π ≈ 28.27 м^2$

$S_{ствола} = πr_0^2 = 0.09π ≈ 0.283 м^2$

$S_{человека} = πr^2 = 0.25π ≈ 0.785 м^2$

$d + r = 1.5 < 3$ => человек находится внутри кроны

* Рассмотрим событие А - яблоко упало на человека

$P(A) = \frac{S_{человека}}{S_{кроны}} = \frac{0.25π}{9π} = \frac{0.25}{9} = \frac{0.25 * 4}{9 * 4} = \frac{1}{36} = 0.0278$

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Circle
import numpy as np

def apple(size):
    R = 3
    r = 0.5
    r0 = 0.3
    d = 1

    fig, ax = plt.subplots(figsize=(6, 6))
    ax.add_patch(Circle((0, 0), R, facecolor='green'))
    ax.add_patch(Circle((0, 0), r0, facecolor='brown'))
    ax.add_patch(Circle((d, 0), r, facecolor='red'))

    X, Y = [], []
    while len(X) < size:
        x = np.random.uniform(-R, R)
        y = np.random.uniform(-R, R)
        if x**2 + y**2 <= R**2 and x**2 + y**2 > r0**2:
            X.append(x)
            Y.append(y)
    X, Y = np.array(X), np.array(Y)

    plt.scatter(X, Y, marker='x', c='black')

    inside = len([1 for x, y in zip(X, Y) if (x - d)**2 + y**2 <= r**2])
    prob = inside / size

    plt.show()
    print(f'На человека из {size} яблок упало {inside}. Доля попавших составила {prob:.4f}')

apple(1000)
```
